```
#configurar API do openrouteservice
api_key = 'eyJvcmciOiI1YjNjZTM1OTc4NTExMTAwMDFjZjYyNDgiLCJpZCI6ImVkZGJmNmRmZWJhNDRkODY4MTVhMGIzNmJlYjY0MWE4IiwiaCI6Im11cm11cjY0In0='
clnt = client.Client(key=api_key)
```

```
# Pede e geocodifica os dois endereços
geolocator = Nominatim (user_agent = 'helenapaulineschulze@gmail.com')
endereco1 = input ("Entre com o primeiro endereço: ")
location1= geolocator.geocode (endereco1)

endereco2 = input ("Entre com o segundo endereço: ")
location2 = geolocator.geocode (endereco2)
```

```
#configurações para o mapa
mapa = folium.Map (location= [location1.latitude, location1.longitude],zoom_start=13)
folium. Marker (location=[location1.latitude, location1.longitude], popup=endereco1,
                 icon=folium.Icon(color='beige',icon='home', prefix='fa')).add_to (mapa)
folium. Marker (location=[location2.latitude, location2.longitude], popup=endereco2,
                 icon=folium.Icon(color='lightgreen',icon='briefcase', prefix='fa')).add_to (mapa)
```

```
# Requisição da Rota
request_params = {'coordinates': [[location1.longitude, location1.latitude],
                                 [location2.longitude, location2.latitude]],
                'format_out': 'geojson',
                'profile': 'driving-car',
                'preference': 'shortest',
                'instructions': 'false',}
route_normal = clnt.directions(**request_params)
```

```
#configurações para a rota
style_route = {'color': '#D9384B'}
```

```
#Adiciona a rota no mapa
folium.features.GeoJson(data=route_normal,
                        name='Rota entre os pontos',
                        style_function=lambda x: style_route,
                        overlay=True).add_to(mapa)
mapa
```
